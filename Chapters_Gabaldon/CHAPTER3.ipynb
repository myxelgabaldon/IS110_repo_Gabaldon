Exercise 2

Write your salary computation to give an employee 1.5 times the hourly rate for hours worked above 40 hours. Example:

    Enter Hours: 45
    Enter Rate: 10
    Pay: 475.0
    
[PEW] E3.2

````python
work_hours = float(input("Enter Hours:"))
rateperhour = float(input("Enter rate:"))

if work_hours > 40:
    regular_hours = 40
    overtime_hours = work_hours - regular_hours
    pay = (regular_hours * rateperhour) + (overtime_hours * rateperhour * 1.5)
    print("Pay:", pay)
elif work_hours >= 0:
    pay = work_hours * rateperhour
    print("Pay:", pay)
else:
    print("Error: Hours cannot be negative.")
````

CHALLENGE

Write a program to prompt for a score between 0 and 100 and print a grade using the following table:

    Score           Grade
    [ 98, 100]      1.0
    [ 94,  97)		1.25
    [ 90,  93)		1.5
    [ 86,  89)		1.75
    [ 82,  85)		2.0
    [ 78,  81)		2.25
    [ 75,  77)		2.5
    [ 0,   74)		5.0

Adapted from [PEW] E3.3
<br>

````python
stud_score = float(input("Enter grades:"))

if 0 <= stud_score <= 100:
    if 98 <= stud_score <= 100:
        grade = 1.0
    elif 94 <= stud_score < 98:
        grade = 1.25
    elif 90 <= stud_score < 94:
        grade = 1.5
    elif 86 <= stud_score < 90:
        grade = 1.75
    elif 82 <= stud_score < 86:
        grade = 2.0
    elif 78 <= stud_score < 82:
        grade = 2.25
    elif 75 <= stud_score < 78:
        grade = 2.5
    elif 0 <= stud_score < 75:
        grade = 5.0
    print("Grade:", grade)
else:
    print("BETTER LUCK NEXT TIME!")
````
